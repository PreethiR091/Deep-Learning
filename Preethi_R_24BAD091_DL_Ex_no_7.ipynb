EXPT NO: 7 - TEXT PREPROCESSING AND EMBEDDING GENERATION USING NLP LIBRARIES

PART A: DATASET PREPARATION AND TEXT PREPROCESSING

Perform the following tasks:

    •	Load the Brown Corpus available in NLTK and select a suitable text category.
    •	Examine sample sentences from the selected dataset.
    •	Convert the text into lowercase.
    •	Perform word tokenization using NLTK.
    •	Remove punctuation and stop words.
    •	Perform lemmatization using WordNetLemmatizer.
    •	Display the original and preprocessed text.



```python
# EXPT NO: 7 - PART A - DATASET PREPARATION AND TEXT PREPROCESSING
# Name : Preethi R
# Roll No: 24BAD091

import nltk
import string
import re
from nltk.corpus import brown, stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

nltk.download('brown')
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

# Load the Brown Corpus news category
original_sentences = [
    ' '.join(sentence)
    for sentence in brown.sents(categories='news')
]

print("\nPreethi R 24BAD091")
print("Total sentences:", len(original_sentences))

print("\nOriginal sample sentences:")
for sentence in original_sentences[:5]:
    print(sentence)

# Initialize preprocessing tools
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

# Preprocess each sentence
preprocessed_sentences = []

for sentence in original_sentences:
    sentence = sentence.lower()
    tokens = word_tokenize(sentence)

    cleaned_tokens = [
        word for word in tokens
        if word.isalpha() and word not in stop_words
    ]

    lemmatized_tokens = [
        lemmatizer.lemmatize(word)
        for word in cleaned_tokens
    ]

    if lemmatized_tokens:
        preprocessed_sentences.append(lemmatized_tokens)

print("\nPreprocessed sample sentences:")
for sentence in preprocessed_sentences[:5]:
    print(' '.join(sentence))

print("\nOriginal sentence:")
print(original_sentences[0])

print("\nPreprocessed sentence:")
print(' '.join(preprocessed_sentences[0]))
```

PART B: VOCABULARY GENERATION AND WORD FREQUENCY ANALYSIS

Perform the following tasks:

    •	Create a vocabulary from the preprocessed text.
    •	Calculate word frequencies using NLTK's FreqDist.
    •	Identify the ten most frequently occurring words.
    •	Display the vocabulary size and frequency distribution.
    •	Prepare the tokenized sentences for Word2Vec training.


```python
# EXPT NO: 7 - PART B - VOCABULARY GENERATION AND WORD FREQUENCY ANALYSIS
# Name : Preethi R
# Roll No: 24BAD091

from nltk import FreqDist

# Flatten all preprocessed sentences into one list
all_words = [
    word
    for sentence in preprocessed_sentences
    for word in sentence
]

# Generate vocabulary and frequency distribution
vocabulary = set(all_words)
frequency_distribution = FreqDist(all_words)

print("Preethi R 24BAD091")
print("Total word tokens:", len(all_words))
print("Vocabulary size:", len(vocabulary))

print("\nTop 10 most frequent words:")
for word, frequency in frequency_distribution.most_common(10):
    print(word, ":", frequency)

print("\nFrequency distribution:")
print(frequency_distribution)

# Prepare tokenized sentences for Word2Vec
word2vec_sentences = preprocessed_sentences

print("\nNumber of sentences for Word2Vec:",
      len(word2vec_sentences))

print("\nSample tokenized sentences:")
for sentence in word2vec_sentences[:3]:
    print(sentence)
```

PART C: WORD EMBEDDING GENERATION USING WORD2VEC

Perform the following tasks:

    •	Train a Word2Vec model using Gensim on the preprocessed sentences.
    •	Use the Skip-gram architecture with an embedding dimension of 100 and a context-window size of 5.
    •	Generate embedding vectors for selected words.
    •	Display the embedding vector and its dimensions.
    •	Identify the five most similar words for a selected word using the trained model.


```python
# EXPT NO: 7 - PART C -WORD EMBEDDING GENERATION USING WORD2VEC
# Name : Preethi R
# Roll No: 24BAD091

!pip install -q gensim

from gensim.models import Word2Vec
import numpy as np

# Train the Skip-gram Word2Vec model
model = Word2Vec(
    sentences=word2vec_sentences,
    vector_size=100,
    window=5,
    min_count=2,
    sg=1,
    workers=4,
    seed=42,
    epochs=10
)

print("Preethi R 24BAD091")
print("Word2Vec model trained successfully.")

print("\nEmbedding dimension:", model.wv.vector_size)
print("Vocabulary size in Word2Vec:", len(model.wv))

# Select a word that exists in the trained vocabulary
selected_word = 'government'

if selected_word not in model.wv:
    selected_word = max(
        model.wv.key_to_index,
        key=lambda word: len(word)
    )

# Generate the embedding vector
embedding_vector = model.wv[selected_word]

print("\nSelected word:", selected_word)
print("Embedding vector:")
print(embedding_vector)

print("\nVector dimensions:", embedding_vector.shape)
print("Number of dimensions:", len(embedding_vector))

# Display the five most similar words
print("\nFive most similar words:")
for word, score in model.wv.most_similar(
    selected_word, topn=5
):
    print(word, ":", round(score, 4))
```


PART D: WORD SIMILARITY ANALYSIS AND EXPERIMENT MANAGEMENT

Perform the following tasks:

    •	Select three meaningful word pairs from the trained vocabulary.
    •	Calculate the cosine similarity between each pair using the Word2Vec model.
    •	Compare the similarity scores and identify the most similar pair.
    •	Analyze the relationship between contextual similarity and the generated word embeddings.
    •	Document the observations and upload the completed notebook to GitHub.



```python
# EXPT NO: 7 - PART D - WORD SIMILARITY ANALYSIS AND EXPERIMENT MANAGEMENT
# Name : Preethi R
# Roll No: 24BAD091

from itertools import combinations

# Candidate pairs with potentially related meanings or contexts
candidate_pairs = [
    ('government', 'state'),
    ('people', 'man'),
    ('new', 'american'),
    ('city', 'state'),
    ('president', 'government'),
    ('country', 'state')
]

# Keep only pairs whose words exist in the trained vocabulary
valid_pairs = [
    (word1, word2)
    for word1, word2 in candidate_pairs
    if word1 in model.wv and word2 in model.wv
]

# If needed, generate fallback pairs from frequent vocabulary words
if len(valid_pairs) < 3:
    frequent_words = [
        word for word, count in frequency_distribution.most_common(100)
        if word in model.wv
    ]

    for word1, word2 in combinations(frequent_words, 2):
        pair = (word1, word2)
        if pair not in valid_pairs and (word2, word1) not in valid_pairs:
            valid_pairs.append(pair)

        if len(valid_pairs) >= 3:
            break

# Use three pairs
selected_pairs = valid_pairs[:3]

print("Preethi R 24BAD091")
print("Cosine Similarity Analysis\n")

similarity_results = []

for word1, word2 in selected_pairs:
    score = model.wv.similarity(word1, word2)
    similarity_results.append((word1, word2, score))
    print(
        f"{word1} - {word2}: {score:.4f}"
    )

if similarity_results:
    most_similar_pair = max(
        similarity_results, key=lambda item: item[2]
    )

    print("\nMost similar pair:")
    print(
        f"{most_similar_pair[0]} - {most_similar_pair[1]} "
        f"with similarity {most_similar_pair[2]:.4f}"
    )

    print("\nObservation:")
    print(
        "A higher cosine similarity indicates that the word vectors "
        "point in more similar directions in the learned vector space. "
        "The scores reflect the contexts learned from the Brown Corpus."
    )
else:
    print("No valid word pairs were found.")
```
